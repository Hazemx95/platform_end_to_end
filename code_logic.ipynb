```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
import pyspark.sql.functions as fn
from pyspark.sql import SparkSession
from pyspark.sql import DataFrame
```

```python
# --- Spark runtime setup: must run BEFORE creating a SparkSession ---
import os, sys, glob, shutil, subprocess


def _find_java_home():
    """Locate a JDK home (Spark 4 needs Java 17 or 21)."""
    jh = os.environ.get("JAVA_HOME")
    if jh and os.path.exists(os.path.join(jh, "bin", "java")):
        return jh
    java = shutil.which("java")
    if java:
        return os.path.dirname(os.path.dirname(os.path.realpath(java)))
    for pattern in (
        os.path.expanduser("~/.local/jdk/jdk-21*"),
        os.path.expanduser("~/.local/jdk/jdk-17*"),
        "/usr/lib/jvm/java-21-openjdk*",
        "/usr/lib/jvm/java-17-openjdk*",
    ):
        hits = sorted(glob.glob(pattern))
        if hits:
            return hits[0]
    raise RuntimeError(
        "No JDK found. Install one (Java 17 or 21), then restart the kernel:\n"
        "    sudo apt-get install -y openjdk-21-jdk-headless"
    )


os.environ["JAVA_HOME"] = _find_java_home()
os.environ["PATH"] = os.path.join(os.environ["JAVA_HOME"], "bin") + os.pathsep + os.environ["PATH"]

# Point driver and workers at this venv's interpreter; PySpark otherwise
# defaults to `python3`, which may not have pyspark installed.
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

print("JAVA_HOME =", os.environ["JAVA_HOME"])
print(subprocess.run(["java", "-version"], capture_output=True, text=True).stderr.strip())
```

```python
spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("CustomerBronzeBatch")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")
```

```python
# Verify the session really works end to end
print(spark.version)

# JVM-only action
print("count:", spark.range(1000).count())

# Action that spawns a Python worker (catches PYSPARK_PYTHON mismatches)
print("udf sum:", spark.range(10).rdd.map(lambda r: r.id).sum())
```
